```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("/content/diabetes.csv")

print(df.head())
```

```python
import seaborn as sns

# 1. Outcome Count
plt.figure()
sns.countplot(x='Outcome', data=df)
plt.title("Diabetes Outcome Count")
plt.show()
```

```python
# 2. Glucose Distribution
plt.figure()
sns.histplot(df['Glucose'], bins=20, kde=True)
plt.title("Glucose Distribution")
plt.show()
```

```python
# 3. BMI vs Age
plt.figure()
sns.scatterplot(x='BMI', y='Age', data=df)
plt.title("BMI vs Age")
plt.show()
```

```python
sns.countplot(x='Outcome', data=df, palette='Set2')


plt.show()
```

```python
sns.histplot(df['Glucose'], bins=20, kde=True, color='orange')


plt.show()
```

```python
sns.scatterplot(x='BMI', y='Age', data=df, color='black')
plt.show()
```
